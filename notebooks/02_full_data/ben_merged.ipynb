```python
# libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# data
df_orders = pd.read_csv('../data/raw/orders.csv')
df_products = pd.read_csv('../data/raw/products.csv')
df_aisles = pd.read_csv('../data/raw/aisles.csv')
df_departments = pd.read_csv('../data/raw/departments.csv')
df_order_products_prior = pd.read_csv('../data/raw/order_products__prior.csv')
df_order_products_train = pd.read_csv('../data/raw/order_products__train.csv')
```

```python
# inspect
print(df_order_products_train.head())
# df_orders.head()
# df_products.head()
# df_aisles.head()
# df_departments.head()
```

## Merge and join dfs

```python
# get sizes of dfs
print(df_order_products_prior.shape)
print(df_order_products_train.shape)
```

```python
# Merge
df_merged = pd.concat([df_order_products_prior, df_order_products_train], ignore_index=True)
df_merged.tail()
```

```python
# check everything was preserved
df_merged.shape
```

```python
df_merged = (
    df_merged
        .merge(df_orders, how="left", on="order_id")
        .merge(df_products, how="left", on="product_id")
        .merge(df_aisles, how="left", on="aisle_id")
        .merge(df_departments, how="left", on="department_id")
)

df_merged.head(10)
```

```python
# Save data as final
df_merged.to_parquet("../data/final.parquet", index=False)
```
